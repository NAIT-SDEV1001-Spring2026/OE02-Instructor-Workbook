```python
math_grades = grades[grades['Class'] == 'Math']
math_grades
```

```python
top_math = math_grades.sort_values(by='Grade', ascending=False)
top_math
```

```python
avg_by_class = grades.groupby('Class')['Grade'].mean()
print(avg_by_class)
```

```python
count_by_class = grades['Class'].value_counts()
print(count_by_class)
```

```python
import matplotlib.pyplot as plt
avg_by_class.plot(kind='bar', title='Average Grade by Class')
plt.ylabel('Average Grade')
plt.show()
```

```python
science_grades = grades[grades['Class'] == 'Science']
top_science = science_grades.sort_values(by='Grade', ascending=False)
print(top_science.iloc[0])
```

```python
grades_below_80 = grades[grades['Grade'] > 80]
print(grades_below_80['Name'].to_string(index=False))
```

```python
print('\n'.join(f"- {name}" for name in grades_below_80['Name']))
```
