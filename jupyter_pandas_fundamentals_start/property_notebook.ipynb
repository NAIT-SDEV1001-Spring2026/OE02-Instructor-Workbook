```python
import pandas as pd
property_data = pd.read_csv('data/Property_Assessment_Data_(Current_Calendar_Year)_20260729.csv')
property_data.head()
```

```python
property_data.shape
```

```python
property_data.describe()
```

```python
property_data["Neighbourhood"].unique()
```

```python
downtown_properties = property_data[property_data['Neighbourhood'] == 'DOWNTOWN']
print(f'Number of properties in downtown Edmonton {downtown_properties.shape[0]}')
```

```python
sorted_downtown = downtown_properties.sort_values(by='Assessed Value', ascending=False)
sorted_downtown[['House Number', 'Street Name', 'Assessed Value']].head(10).style.format({'House Number': '{:.0f}', 'Assessed Value': '${:,.0f}'})
```

```python
avg_by_neighbourhood = property_data.groupby('Neighbourhood')['Assessed Value'].agg(['mean', 'median', 'count'])
avg_by_neighbourhood.sort_values('mean', ascending=False).head(10).style.format({
    'mean': '${:,.0f}', 'median': '${:,.0f}', 'count': '{:.0f}'
})
```
